```python
#Getting Data from yahoo finance
import yfinance as yf 

data = yf.download("QQQ", period="1y", interval="1d")
data = data.dropna()
```

```python
data["return"] = data["Close"].pct_change()
data["target"] = (data["return"].shift(-1) > 0).astype(int)
```
